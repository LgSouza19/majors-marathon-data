```python
def baixar_boston_completo(
    ano_inicio=2001, ano_fim=2026, pasta_destino="data"
):
    os.makedirs(pasta_destino, exist_ok=True)
    todos_os_anos = []

    print(
        f"🚀 INICIANDO INGESTÃO HISTÓRICA: Boston Marathon ({ano_inicio} - {ano_fim})"
    )

    for ano in range(ano_inicio, ano_fim + 1):
        # 1. Checa a quantidade exata de corredores na edição
        check_params = {
            "select": "ID",
            "ShortDescription": f"ilike.*{ano} Boston Marathon*",
            "ReportingSegment": "in.(Runners,runner)",
            "limit": "1",
        }

        try:
            init_res = requests.get(
                BASE_URL, headers=HEADERS, params=check_params, timeout=10
            )
            content_range = init_res.headers.get("content-range", "")
            total_atletas = (
                int(content_range.split("/")[-1])
                if "/" in content_range
                else None
            )

            if not total_atletas:
                print(f"⚪ Ano {ano}: Nenhum registro encontrado.")
                continue

            print(
                f"\n📥 Ano {ano}: Baixando {total_atletas:,} corredores com todas as colunas..."
            )

            # 2. Paginação em lotes de 1.000
            limit = 1000
            total_batches = math.ceil(total_atletas / limit)
            registros_ano = []

            for batch in tqdm(range(total_batches), desc=f"Boston {ano}"):
                offset = batch * limit
                params = {
                    "select": "*",  # TODAS as colunas disponíveis
                    "ShortDescription": f"ilike.*{ano} Boston Marathon*",
                    "ReportingSegment": "in.(Runners,runner)",
                    "order": "RankOverAll.asc.nullslast",
                    "offset": str(offset),
                    "limit": str(limit),
                }

                res = requests.get(
                    BASE_URL, headers=HEADERS, params=params, timeout=15
                )
                if res.status_code in (200, 206):
                    dados = res.json()
                    if not dados:
                        break
                    registros_ano.extend(dados)
                else:
                    print(f"\n❌ Erro {res.status_code} no lote {batch}: {res.text}")
                    break

                time.sleep(0.05)  # Respeito ao servidor

            df_ano = pd.DataFrame(registros_ano)
            # Salva o backup individual do ano em parquet
            df_ano.to_parquet(
                f"{pasta_destino}/boston_{ano}.parquet", index=False
            )
            todos_os_anos.append(df_ano)

        except Exception as e:
            print(f"\n❌ Falha ao processar ano {ano}: {e}")

    # 3. Consolidação final
    if todos_os_anos:
        df_completo = pd.concat(todos_os_anos, ignore_index=True)
        caminho_final = f"{pasta_destino}/boston_marathon_2001_2026_raw.parquet"
        df_completo.to_parquet(caminho_final, index=False)
        print(f"\n" + "=" * 60)
        print(
            f"🎉 SUCESSO! Base consolidada salva em: {caminho_final}"
        )
        print(f"📊 Total de linhas: {len(df_completo):,}")
        print(f"📊 Total de colunas: {df_completo.shape[1]}")
        print("=" * 60)
        return df_completo
    return pd.DataFrame()


# Execução do download completo
df_boston_all = baixar_boston_completo(2001, 2005)
```
