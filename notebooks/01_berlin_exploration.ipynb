```python
import time
import pandas as pd
import requests

BERLIN_API_URL = "https://api.results.scc-events.com/result"

BERLIN_HEADERS = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.15; rv:153.0) Gecko/20100101 Firefox/153.0",
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Origin": "https://results.scc-events.com",
    "Referer": "https://results.scc-events.com/",
}

print("🔍 Varrendo anos disponíveis na API de Berlim (1990 a 2026)...\n")

anos_berlim = []

for ano in range(1990, 2027):
    params = {
        "ek": "BM",
        "ci": "BML",
        "y": str(ano),
        "t": f"BM_{ano}",
        "draw": "1",
        "start": "0",
        "length": "1",
    }

    try:
        r = requests.get(
            BERLIN_API_URL, headers=BERLIN_HEADERS, params=params, timeout=5
        )

        if r.status_code == 200:
            data = r.json()
            total = data.get("recordsTotal", 0)

            if total and total > 0:
                anos_berlim.append({"Ano": ano, "Total_Atletas": total})
                print(f"✅ Ano {ano}: {total:,} maratonistas")
            else:
                print(f"⚪ Ano {ano}: 0 registros")
        else:
            print(f"❌ Ano {ano}: Status HTTP {r.status_code}")

    except Exception as e:
        print(f"⚠️ Erro no ano {ano}: {e}")

    time.sleep(0.05)

# Resumo em DataFrame
df_resumo_berlim = pd.DataFrame(anos_berlim)
print("\n" + "=" * 50)
if not df_resumo_berlim.empty:
    print(
        f"🏆 Período Coberto: {df_resumo_berlim['Ano'].min()} a {df_resumo_berlim['Ano'].max()}"
    )
    print(f"📊 Total de edições disponíveis: {len(df_resumo_berlim)}")
    print(
        f"🏃 Volume Total de Maratonistas em Berlim: {df_resumo_berlim['Total_Atletas'].sum():,}"
    )
print("=" * 50)
```

```python
import json
import requests

BERLIN_API_URL = "https://api.results.scc-events.com/result"

BERLIN_HEADERS = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.15; rv:153.0) Gecko/20100101 Firefox/153.0",
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Origin": "https://results.scc-events.com",
    "Referer": "https://results.scc-events.com/",
}

ano_teste = 2019

# Variações comuns de identificador da SCC Events / Mika Timing
variacoes_ci = ["BML", "MAL", "BM", "L", "ML", "MAL_Lauf", "BM_Lauf"]
variacoes_t = [
    f"BM_{ano_teste}",
    f"MAL_{ano_teste}",
    f"BM",
    f"MAL",
    f"BML_{ano_teste}",
    f"{ano_teste}",
]

print(
    f"🔍 Testando combinações de parâmetros para recuperar Berlim {ano_teste}...\n"
)

encontrado = False
for ci in variacoes_ci:
    for t in variacoes_t:
        params = {
            "ek": "BM",
            "ci": ci,
            "y": str(ano_teste),
            "t": t,
            "draw": "1",
            "start": "0",
            "length": "1",
        }
        try:
            r = requests.get(
                BERLIN_API_URL, headers=BERLIN_HEADERS, params=params, timeout=3
            )
            if r.status_code == 200:
                data = r.json()
                total = data.get("recordsTotal", 0)
                if total > 1000:
                    print(
                        f"🎯 SUCESSO ENCONTRADO! ci='{ci}', t='{t}' -> {total:,} maratonistas!"
                    )
                    encontrado = True
                    break
        except Exception:
            pass
    if encontrado:
        break

if not encontrado:
    print(
        "Nenhuma combinação simples bateu. Vamos testar o endpoint legados da Mika Timing."
    )
```

```python
import json
import pandas as pd
import requests

BERLIN_API_URL = "https://api.results.scc-events.com/result"
BERLIN_HEADERS = {
    "User-Agent": "Mozilla/5.0",
    "Origin": "https://results.scc-events.com",
    "Referer": "https://results.scc-events.com/",
}

# Puxa uma amostra variada de 2024 e 2019
for ano, ci in [(2024, "BML"), (2019, "MAL")]:
    params = {
        "ek": "BM",
        "ci": ci,
        "y": str(ano),
        "t": f"BM_{ano}",
        "draw": "1",
        "start": "0",
        "length": "500",
    }
    res = requests.get(BERLIN_API_URL, headers=BERLIN_HEADERS, params=params)
    df_check = pd.DataFrame(res.json().get("data", []))

    print(f"--- DIAGNÓSTICO BERLIM {ano} (ci='{ci}') ---")
    if "description" in df_check.columns:
        print("Tipos de Prova (description):")
        print(df_check["description"].value_counts(dropna=False))
    if "ident" in df_check.columns:
        print("\nIdentificadores (ident):")
        print(df_check["ident"].value_counts(dropna=False))
    print("-" * 50)
```
